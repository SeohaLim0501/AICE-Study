```python
# 판다스에서 대표적으로 사용하는 자료구조: 시리즈와 데이터프레임
# 시리즈: 변수명 = pd.Series(data, index = index, name = name)
s = pd.Series(['대한민국', '포르투갈','가나', '우루과이'], index = ['가', '나', '다', '라'], name = "2022 카다르월드컵 H조")
print(s)
```

```python
# read_csv 함수의 파라미터를 활용해 데이터프레임 정제

flight2 = pd.read_csv('./Clean_Dataset.csv', index_col = 'stops', usecols = ['stops', 'departure_time', 'arrival_time', 'destination_city'])
flight2

# index_col은 DataFrame의 Index로 사용할 CSV의 열을 지정하는 매개변수
# Column position으로도 지정 가능: ex) index_col = 3

# usecols는 반드시 index_col로 설정한 칼럼명을 포함해야 함
```

```python
# crosstab는 범주형 데이터 2개를 비교, 분석할 때 사용
# 두 변수의 조합별 데이터 개수를 세는 함수

pd.crosstab(index = flight.source_city, columns = flight.arrival_time)
```

```python
# 개수 대신 비율을 출력할 경우 ex) normalize = 'index'
# 이 경우 index 별로 column 값들의 합이 1이 됨; 각 index의 합계로 나눔

pd.crosstab(index = flight.source_city, columns = flight.arrival_time, normalize = 'index')
```

```python
# 전체 합계까지 출력할 경우 margins = True

pd.crosstab(index = flight.source_city, columns = flight.arrival_time, margins = True)
```

```python
# 둘 다 사용할 경우 normalized 된 행/열은 margin 생략

pd.crosstab(index = flight.source_city, columns = flight.arrival_time, normalize = 'index', margins = True)
```
