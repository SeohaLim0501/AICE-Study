## 조건으로 데이터 선택하기

```python
flight = pd.read_csv('./Clean_Dataset.csv', encoding = "cp949")
flight
```

```python
flight_extract = flight[(flight['price']>12000) & (flight['airline'] == 'Air_India')]
flight_extract
```

```python
# 조건을 변수로 저장하여 사용할 수도 있다.

price_Tag = flight['price'] > 12000
airline_Tag = flight['airline'] == 'Air_India'

'''
위에서 실제로 일어나는 일

flight['price']: flight DataFrame에서 price 열만 꺼내서 Series를 만듦
flight['price'] > 12000: NumPy 배열과 똑같이 원소별 비교를 통해 Boolean Series를 만듦
'''

'''
두 Boolean Series에 &를 하면 element-wise operation이 일어남
'''

flight[price_Tag & airline_Tag]
```

## 정리
중요한 차이가 있다.  
Column을 선택할 때는 flight[] 안에 str, list가 들어간다.  
Row를 선택할 때는 flight[] 안에 slice, Boolean Series가 들어간다. 이때 Boolean Series는 Pandas Series 타입인데 각 원소가 전부 Boolean인 Series이다.  
실제로 row를 선택하고 싶은데 flight[] 안에 row name (인덱스)를 넣으면 뜻하지 않은 결과가 초래한다.

```python
flight[1]
```

참고로 flight[] 안에는 Boolean Series 뿐만 아니라 다른 종류도 Series도 들어갈 수 있다.  
  
1. Boolean Series: 행 필터링  
ex) S = pd.Series([True, False, True, False, True])  
flight[S]: True에 해당하는 행만 선택  
=> 이때 S의 indexing이랑 flight의 indexing이랑 일치해야 한다. 예를 들어 flight를 arange로 다르게 indexing을 했다면 S에서 인덱싱을 이에 맞춰서 하지 않으면 오류가 발생한다.  
  
2. String Series: 열 필터링
ex) S = pd.Series(['airline', 'price'])
flight[S]

## 참고: Column Name 중복 문제  

```python
df = pd.DataFrame(
    [[1, 2, 3]],
    columns = ['a', 'a', 'b']
)

print(df['a'])
```

이때 흥미로운 문제가 발생한다.  
df['b']는 정상적으로 Series를 반환하는데, df['a']는 DataFrame을 반환한다.  
여기서 매우매우 중요한 차이가 나온다.  
df[] 안에 Boolean Series가 들어가면 기본적으로 False에 해당하는 행은 지워진다. 그러나 df[] 안에 Boolean DataFrame이 들어가면 False에 해당하는 값은 NaN으로 마스킹한다. 이유는 Series는 1차원이라 행에 일대일 대응 시킬 수 있는데 DataFrame은 2차원이라서 각 값에 대응되기 때문에 지우면 DataFrame에 구멍을 내는 것이 되어버린다..  
  
이 얘기가 왜 나오냐,  
df[df['a']>10] 이면 속속이 NaN이 나오는 구멍 뚫린 DataFrame이 나오지만  
df[df['b']>10] 이면 깔끔하게 행이 지워진 DataFrame이 나온다.

```python
# 중복을 금지할 수도 있음

df2 = pd.DataFrame(
    [[1, 2, 3]],
    columns = ['a', 'b', 'c']
)
df2 = df2.set_flags(allows_duplicate_labels=False)

# 현재 Column Name이 중복되어 있는지 확인

print(df.columns.is_unique)
print(df2.columns.is_unique)
```

# Section 03: 데이터프레임 변형하기

```python
import pandas as pd
import numpy as np

flight = pd.read_csv('./Clean_Dataset.csv', encoding = "cp949")
flight
```

## 그룹화하기

### 1) groupby 활용하기

```python
# 데이터프레임의 같은 값을 하나로 묶어서 통계나 집계 결과를 확인하기 위해 사용

# 1. Split: 그룹별로 데이터를 나누기
# 2. Apply: 각 그룹별로 집계 함수를 적용
# 3. Combine: 그룹별 집계 결과를 하나로 합침

airline_group = flight.groupby('airline')
airline_group
```

```python
airline_group.groups
```

```python
# 가독성을 위해 내부 메소드 활용

airline_group.count()
```

```python
# .min() 메소드는 항공사별로 각 열의 최솟값을 계산

airline_group.min()
```

```python
airline_group.mean(numeric_only = True)

# numeric type이 아닌 datetime, timedelta 같은 dtype도 평균을 계산할 수 있음
```

```python
airline_group.mean(numeric_only = True)[['price']]

# pandas 2.0부터 Groupby.mean()의 numeric_only 기본값이 False로 변경
```

```python
# 복수 개의 칼럼을 기준으로 groupby 적용

flight.groupby(['airline', 'arrival_time']).mean(numeric_only=True)
```

```python
# groupby 후 원하는 데이터만 가져오기

flight.groupby(['airline', 'arrival_time']).mean(numeric_only = True).loc[[('AirAsia', 'Evening')]]
```

### 2) 인덱스로 그룹화하기

```python
# set_index()로 인덱스 설정하기

flight.set_index(['airline', 'arrival_time'])
```

```python
flight.set_index(['airline', 'arrival_time']).groupby(level=[0]).mean(numeric_only=True)
```

```python
flight.set_index(['airline', 'arrival_time']).groupby(level=[0,1]).mean(numeric_only=True)
```

### 3) Aggregate로 집계하기

```python
# 한 번에 최솟값, 최댓값 확인

flight.set_index(['airline', 'arrival_time']).groupby(level=[0,1]).aggregate([np.min,np.max])
```

## 피벗테이블 생성하기

```python
pivot_data = pd.DataFrame({'cust_id': ['cust_1', 'cust_1', 'cust_1', 'cust_2', 'cust_2', 'cust_2', 'cust_3', 'cust_3', 'cust_3'],
                          'prod_cd': ['p1', 'p2', 'p3', 'p1', 'p2', 'p3', 'p1', 'p2', 'p3'],
                          'grade': ['A', 'A', 'A', 'A', 'A', 'A', 'B', 'B', 'B'],
                          'purch_amt': [30, 10, 0, 40, 15, 30, 0, 0, 10]})
pivot_data
```

```python
# pivot은 index, columns, values를 지정하여 구성
# index를 행으로, columns를 열 attribute로 사용하여 각 순서쌍마다 values에 있는 값을 대응
# 대응하는 값이 없는 경우 NaN으로 표기
# 대응하는 값이 하나 이상인 경우 에러

pivot_data.pivot(index = 'cust_id', columns = 'prod_cd', values = 'purch_amt')
```

```python
# 만약 values를 생략한다면 index와 columns로 지정하지 않은 나머지 모든 컬럼을 값으로 사용
pivot_data.pivot(index = 'cust_id', columns = 'prod_cd')

# 이때 MultiIndex가 생성되며, 가장 높은 레벨이 values의 컬럼명, 낮은 레벨이 columns의 값이 됨
```

```python
# pivot_table은 pivot과 달리 대응되는 값이 여러 개여도 aggfunc를 이용해 하나로 합쳐서 표시할 수 있음
# aggfunc의 기본값은 np.mean

pivot_data.pivot_table(index='grade', columns='prod_cd', values='purch_amt', aggfunc=np.sum)

# pivot_table은 pivot과 달리 인자 순서가 다름
# .pivot_table(values, index, columns, aggfunc)
```

```python
pivot_data.pivot_table('purch_amt', 'grade', 'prod_cd', np.sum)
```

## concat 활용하여 병합하기

```python
df1 = pd.DataFrame({'col1': ['사과', '배', '감', '수박', '메론'], 'col2': [500, 1000, 2500, 5000, 3000]}, index = [0, 1, 2, 3, 4])
df2 = pd.DataFrame({'col1': ['수박', '멜론', '딸기', '키위', '오렌지'], 'col2': [5000, 3000, 1000, 600, 700]}, index = [3 ,4, 5, 6, 7])
df1
```

```python
df2
```

```python
pd.concat([df1, df2], ignore_index = False)
```

```python
pd.concat([df1, df2], ignore_index = True)
```

```python
pd.concat([df1, df2], axis = 1)
```

```python
# column 단위로 합칠 때 ignore_index를 True로 지정해주면 column index의 이름이 0 1 2.. 로 초기화된다

pd.concat([df1, df2], axis = 1, ignore_index = True)
```

```python
df3 = pd.DataFrame({'item': ['item0', 'item1', 'item2', 'item3'], 'count': ['count0', 'count1', 'count2', 'count3'],
                    'price': ['price0', 'price1', 'price2', 'price3']}, index = [0, 1, 2, 3])
df4 = pd.DataFrame({'item': ['item2', 'item3', 'item4', 'item5'], 'count': ['count2', 'count3', 'count4', 'count5'],
                    'price': ['price2', 'price3', 'price4', 'price5'], 'var': ['var2', 'var3', 'var4', 'var5']}, index = [2, 3, 4, 5])
df3
```

```python
df4
```

```python
# outer은 합집합으로 모든 칼럼으로 병합

pd.concat([df3, df4], join = 'outer')
```

```python
# inner은 교집합으로 공통으로 존재하는 칼럼만 병합

pd.concat([df3, df4], join = 'inner')
```
